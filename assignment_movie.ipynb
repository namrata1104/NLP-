```python
import string
import spacy
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from spacy import displacy
import re

```

```python
nlp = spacy.load("en_core_web_sm")
```

```python
movies = [
    {
        "title": "Interstellar",
        "description": "A team of astronauts travel through a wormhole in space to find a new home for humanity."
    },
    {
        "title": "Wall-E",
        "description": "A small waste-collecting robot falls in love while uncovering a plot that threatens humanity on a future Earth."
    },
    {
        "title": "The Martian",
        "description": "An astronaut is stranded on Mars and must use science and ingenuity to survive until rescue arrives."
    },
    {
        "title": "Ratatouille",
        "description": "A rat with a passion for cooking teams up with a young chef in a famous Paris restaurant."
    },
    {
        "title": "Gravity",
        "description": "Two astronauts work together to survive after an accident leaves them stranded in orbit around Earth."
    },
    {
        "title": "Transformers",
        "description": "Giant robots from outer space battle on Earth with humans caught in the middle of the conflict."
    },
    {
        "title": "Chef",
        "description": "A chef who loses his restaurant job starts a food truck serving delicious Cuban sandwiches across America."
    },
    {
        "title": "Ex Machina",
        "description": "A programmer is invited to administer the Turing test to an intelligent humanoid robot with artificial intelligence."
    },
    {
        "title": "Julie and Julia",
        "description": "A young woman cooks all 524 recipes from Julia Child's cookbook and blogs about food and French cuisine."
    },
    {
        "title": "2001 A Space Odyssey",
        "description": "A space voyage to Jupiter turns dangerous when the ship's artificial intelligence HAL 9000 develops a mysterious agenda."
    },
]
```

```python
def preprocess(text):
    text = text.lower()
    text = re.sub(r'[^\w\s]' , '', text)
    doc = nlp(text)
    tokens = [
        token.lemma_
        for token in doc
        if not token.is_stop and not token.is_space
    ]
    return " ".join(tokens)    
processed_description = [preprocess(movie["description"]) for movie in movies]
```

```python
vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(processed_description)
```

```python
query = "space adventure with robots"
processed_query = preprocess(query)
query_vector = vectorizer.transform([processed_query])
```

```python
similarity_scores = cosine_similarity(query_vector, tfidf_matrix).flatten()
```

```python
print("\nCosine Similarity Matrix:\n")
print(similarity_scores)
```

```python
top_3_indices = similarity_scores.argsort()[::-1][:3]

print(f"Query: '{query}'\n")
print("\nTop 3 Results:")
print(top_3_indices)

for rank, idx in enumerate(top_3_indices, start=1):
    print(f"\nRank {rank}:")
    print(f"Title      : {movies[idx]['title']}")
    print(f"Description: {movies[idx]['description']}")
    print(f"Score      : {round(similarity_scores[idx], 4)}")
```
