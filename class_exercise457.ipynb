```python
import spacy
from spacy import displacy
from IPython.display import HTML, display

#   Load the English language model
nlp = spacy.load("en_core_web_sm")

text = "I saw the man with a telescope."

doc = nlp(text)

# Print the dependency parse information for each token
for token in doc:
    print(f"{token.text:<12} {token.dep_:<12} {token.head.text}")

html = displacy.render(doc, style="dep", jupyter=False)

# Display the dependency parse visualization in Jupyter Notebook
display(HTML(html))

# Discuss with your partner:

# 1. What is the ROOT? - "saw" is the ROOT confirmed in the table

# 2. What does "with a telescope" modify? -  with a telescope modifies "saw". by the long "prep" arc from "saw" to "with"

# 3. Does the sentence have more than one meaning? yes it has ,below are the example.
#    I used a telescope to see the man. 
#    I saw the man who had a telescope.

# 4. Did spaCy choose the interpretation you expected?
#No

# 5. Rewrite the sentence to remove ambiguity?-  I saw the man who was carrying a telescope.
```

```python
import spacy

nlp = spacy.load("en_core_web_sm")

text = """
Microsoft announced a new AI research initiative
in Berlin on 15 June 2026.

The project involves collaboration with Google,
SAP and SRH University Heidelberg.

Satya Nadella will present the initiative
at a conference in Germany.
"""

doc = nlp(text)

for ent in doc.ents:
    print(
        ent.text,
        ent.label_,
        spacy.explain(ent.label_)
    )

#1. Which entities were detected? - Microsoft,AI,Berlin,Google,SAP, Satya Nadella, Germany,15 June 2026
#2. Which labels were assigned?- ORG ,GPE Countries,DATE ,PERSON
#3. Which entity types occur most often? - ORG occurs more frequently almost 4 times.institutions occurs 
#4. Were any entities missed?- SRH university heidelberg
```

```python

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

documents = [
    "AI is transforming healthcare",
    "Machine learning improves healthcare",
    "Football is a popular sport"
]

vectorizer = TfidfVectorizer()

tfidf_matrix = vectorizer.fit_transform(documents)

print("Vocabulary:")
print(vectorizer.get_feature_names_out())

print("\nTF-IDF Matrix:")
print(tfidf_matrix.toarray())

similarity_matrix = cosine_similarity(tfidf_matrix)

print("\nCosine Similarity Matrix:")
print(similarity_matrix)

print("Doc1 vs Doc2:", round(similarity_matrix[0][1], 3))
print("Doc1 vs Doc3:", round(similarity_matrix[0][2], 3))
print("Doc2 vs Doc3:", round(similarity_matrix[1][2], 3))

# 1. Which pair has the highest similarity?- Doc1 shares equal similarity with Doc2 and Doc3
# 2. Is this result reasonable? - Doc1 and Doc2 are related
# 3. Why does TF-IDF treat some words as more important? - TF-IDF gives higher weight to words that are Frequent in a document
# 4. Does cosine understand meaning? - cosine similarity only measures vector angle based on shared vocab.
# 5. Would it know that "AI" and "machine learning" are related?-TF-IDF treat "AI" and "machine learning" as different tokens
```
